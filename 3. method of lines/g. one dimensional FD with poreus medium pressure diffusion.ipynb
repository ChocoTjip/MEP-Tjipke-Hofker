# One dimensional dummy problem

In this section the goal will be to solve the 1 dimensional dummy problem for a first step to pressure gradients

\begin{equation}
\begin{aligned}
\rho \partial_t v &=-RT\partial_x\rho-\rho v \partial_x v+\mu\partial_x^2 v -S\\
\partial_t\rho&=-(\partial_x \rho)v-\rho (\partial_x v) + \mu_\epsilon\partial_x^2\rho
\end{aligned}
\end{equation}

Where

\begin{equation}
S = \frac{\mu}{K}v
\end{equation}

under IC:

\begin{equation}
\left\{ \begin{aligned}
\rho(x,0) &= 1 \\
v(x,0) &= 0
\end{aligned}\right.
\end{equation}

and BC:
\begin{equation}
\left\{\begin{aligned}
\rho(0,t) &= 2-e^{-10t} \\
\partial_x v(0,t) &= 0 \\
v(L,t) &= 0
\end{aligned}\right.
\end{equation}

```julia
using OrdinaryDiffEq, MethodOfLines, DomainSets, ModelingToolkit;
```

## Definitions

```julia
# Parameters, variables, and derivatives
@parameters t x
@variables ρ(..) v(..)

# definitions of derivatives
Dt = Differential(t)
Dx = Differential(x)
Dxx = Differential(x)^2

# constants
R = 1.0
T = 1.0
μ = 1.0
μρ = 0.001
K = 1.0

# PDE
eq = [
    Dt(v(t, x)) ~ (-R*T*Dx(ρ(t, x))+μ*Dxx(v(t, x))-ρ(t, x)*v(t, x)*Dx(v(t, x))-v(t, x)*μ/K)/ρ(t, x),
    Dt(ρ(t, x)) ~ -Dx(ρ(t, x))*v(t, x)-ρ(t, x)*Dx(v(t, x))+μρ*Dxx(ρ(t, x))
]

# Boundary conditions
bcs = [
    ρ(0, x) ~ 1.0,
    v(0, x) ~ 0.0,
    ρ(t,0) ~ 2.0 -exp(-10.0*t),
    Dx(v(t, 0)) ~ 0.0,
    v(t, 1) ~ 0.0
]

# Space and time domains
domains = [
    t ∈ Interval(0.0, 8.0),
    x ∈ Interval(0.0, 1.0)
]
```

## PDE

```julia
# PDE system
@named pdesys = PDESystem(
    eq, 
    bcs, 
    domains, 
    [t, x], [ρ(t, x), v(t, x)]
)
```

```julia
# Method of lines discretization
dx = .02
order = 1

MOL_scheme = UpwindScheme(order)
discretization = MOLFiniteDifference([x => dx], t, advection_scheme = MOL_scheme)

# Convert the PDE problem into an ODE problem
prob = discretize(pdesys, discretization);
```

# Solver

```julia
using OrdinaryDiffEq
```

```julia
# Solve ODE problem
sol = solve(prob, Tsit5(), saveat = 0.1)
```

## Plotting/animating

```julia
# Plot results and compare with exact solution
discrete_x = sol[x];
discrete_t = sol[t];
sol_ρ = sol[ρ(t, x)];
sol_v = sol[v(t, x)];


folder = "g. pressure diffusion";
mkpath(folder);
```

```julia
# Used for plotting
using Plots
```

```julia
# limits
ρmin = minimum(sol_ρ)
ρmax = maximum(sol_ρ)

vmin = minimum(sol_v)
vmax = maximum(sol_v)

left_min = ρmin
left_max = ρmax

anim = @animate for k in 1:length(discrete_t)

    # top plot: density + total pressure
    p1 = plot(
        discrete_x,
        sol_ρ[k, :],
        label = "density",
        color = :blue,
        xlabel = "x",
        ylabel = "density",
        title = "t = $(round(discrete_t[k], digits=3))",
        ylims = (0, 4),
        legend = :topright
    )

    # bottom plot: velocity
    p2 = plot(
        discrete_x,
        sol_v[k, :],
        label = "velocity",
        color = :green,
        xlabel = "x",
        ylabel = "velocity",
        ylims = (vmin, vmax),
        legend = :topright
    )

    # combine
    plot(
        p1,
        p2,
        layout = (2,1),
        size = (800,600)
    )

end

gif(anim, joinpath(folder, "flow.gif"), fps = 10)
```
