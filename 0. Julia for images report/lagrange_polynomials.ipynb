\begin{equation}
    \phi^i(\xi)
    =
    \prod_{\substack{j=1 \\ j\neq i}}^{p+1}
    \frac{\xi-\xi_j}{\xi_i-\xi_j}.
\end{equation}

```julia
function phi_i(i, p, ξ, ξ_j)
    product = 1.0
    ξ_i = ξ_j[i]    # Get the interpolation node for index i    
    for j in 1:p+1
        if j != i
            product *= (ξ - ξ_j[j]) / (ξ_i - ξ_j[j])
        end
    end
    return product
end
```

```julia
ξ_j = [-1, -0.1, 0.5, 1]    # Interpolation nodes for a 3rd-degree polynomial
p = length(ξ_j)-1           # Degree of the polynomial
```

```julia
phis = [
    [phi_i(i, p, ξ, ξ_j) for ξ in -1:0.01:1] for i in 1:length(ξ_j)
]
```

```julia
using Plots
```

```julia
p1 = plot(
    -1:0.01:1,
    phis[1],
    label="φ₁(ξ)",
    xlabel="ξ",
    ylabel="φᵢ(ξ)",
    title="Lagrange Basis Functions",
    legend=:bottomright,
)

plot!(
    -1:0.01:1,
    phis[2],
    label="φ₂(ξ)",
)

plot!(
    -1:0.01:1,
    phis[3],
    label="φ₃(ξ)",
)

plot!(
    -1:0.01:1,
    phis[4],
    label="φ₄(ξ)",
)

plot!(
    ξ_j, 
    [0,0,0,0], 
    seriestype=:scatter, 
    label="Interpolation Nodes", 
    color=:black
)
```
