```julia
x = [0, 1,2,3,4,5, 6]
y = [1, 1,1,1,0,0, 0]

xc = [2,4]
yc = [1,0]

xu_p = [2,3]
yu_p = [1,1]

xu_n = [3,4]
yu_n = [1,0]
```

```julia
using Plots
```

```julia
p1 = plot(x, y, seriestype=:scatter, color=:black, label="Data Points", xlabel="x", ylabel="y", title="Derivative central diferences", legend=:topright)
p2 = plot(x, y, seriestype=:scatter, color=:black, label="Data Points", xlabel="x", ylabel="y", title="Derivatives upwind methods", legend=:topright)
plot!(p1, xc, yc, color=:blue, label="Central Difference", markershape=:circle)
plot!(p2, xu_p, yu_p, color=:red, label="Upwind Positive", markershape=:circle)
plot!(p2, xu_n, yu_n, color=:green, label="Upwind Negative", markershape=:circle)

plot(p1, p2, layout=(1, 2), size=(800, 400))
```
